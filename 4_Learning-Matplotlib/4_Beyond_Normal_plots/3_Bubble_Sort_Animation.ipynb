```python
# Import all modules
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
import matplotlib as mp
import numpy as np
import random

# Bubble Sort function
def bubblesort(A):
    n = len(A)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):

            if A[j] > A[j + 1]:
                A[j], A[j + 1] = A[j + 1], A[j]
                swapped = True

            # Yield current array
            yield A.copy()

        if not swapped:
            break

# Function to display animation
def showGraph():

    n = 20
    a = [i for i in range(1, n + 1)]
    random.shuffle(a)

    datasetName = 'Random'
    algoName = 'Bubble Sort'

    # Store all animation frames
    frames = [a.copy()]
    frames.extend(list(bubblesort(a.copy())))

    # Set graph style
    plt.style.use('fivethirtyeight')

    # Set bar colors
    data_normalizer = mp.colors.Normalize(vmin=1, vmax=n)
    color_map = mp.colors.LinearSegmentedColormap.from_list(
        "my_map",
        ["#3B82F6", "#E4365D"]
    )

    fig, ax = plt.subplots(figsize=(10, 6))

    # Create bars
    bar_rects = ax.bar(
        range(len(a)),
        a,
        align="edge",
        color=color_map(data_normalizer(range(1, n + 1)))
    )

    ax.set_xlim(0, len(a))
    ax.set_ylim(0, int(1.1 * len(a)))

    ax.set_title(
        "ALGORITHM : " + algoName + "\n" + "DATA SET : " + datasetName,
        fontdict={
            'fontsize': 13,
            'fontweight': 'medium',
            'color': '#E4365D'
        }
    )

    # Iteration counter
    text = ax.text(
        0.01, 0.95, "",
        transform=ax.transAxes,
        color="#E4365D"
    )

    iteration = [0]

    # Animation function
    def animate(A, rects, iteration):
        for rect, val in zip(rects, A):
            rect.set_height(val)

        iteration[0] += 1
        text.set_text("Steps : {}".format(iteration[0]))

    # Create animation
    anim = FuncAnimation(
        fig,
        func=animate,
        fargs=(bar_rects, iteration),
        frames=frames,
        interval=50,
        repeat=False,
        cache_frame_data=False
    )

    # Display animation in JupyterLab
    display(HTML(anim.to_jshtml()))
    plt.close(fig)

# Run program
showGraph()
```
