````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
````

# Syntax
## matplotlib.pyplot.pie(x, labels=None, colors=None, autopct=None, explode=None, startangle=0, shadow=False)

# Parameters:

- x: Specifies the numerical values used to determine the size of each pie slice.
- labels: Specifies the labels displayed for each slice.
- colors: Specifies the colors of the pie slices.
- autopct: Displays the percentage value on each slice using a specified format (for example, "%1.1f%%").
- explode: Separates one or more slices from the center of the pie chart.
- startangle: Rotates the starting position of the pie chart by the specified angle.
- shadow: If set to True, adds a shadow effect below the pie chart.

````python
y = np.array([35, 25, 25, 15])

plt.pie(y)
plt.show() 
````

````python
labelss = ["A", "B", "C", "D"]
values = [30, 25, 20, 25]

plt.pie(values, labels=labelss)
plt.title("Simple Pie Chart")
plt.show()
````

### Understanding `autopct`

| Format      | Meaning                     |
| ----------- | --------------------------- |
| `'%1.1f%%'` | Percentage with 1 decimal   |
| `'%1.2f%%'` | Percentage with 2 decimals  |
| `'%1.0f%%'` | Percentage with no decimals |

---

````python
y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
mycolors = ["Red", "Yellow", "blue", "grey"]

plt.pie(y, labels = mylabels, colors = mycolors, autopct=lambda p: f'{p:.1f}%\n({p*sum(y)/100:.0f} kg)')
plt.legend(title = "Four Fruits:")
plt.show() 
````

## Understanding the formula

```
lambda p: f'{p:.1f}%\n({p*sum(y)/100:.0f} kg)'
```

- `p` → Calculated percentage.
- `p*sum(y)/100` → Actual value.
- `:.1f` → One decimal place.
- `:.0f` → No decimal places.
- `\n` → Moves the value to the next line.

Tip: Change `kg` to any unit, such as `students`, `₹`, or `litres`.
