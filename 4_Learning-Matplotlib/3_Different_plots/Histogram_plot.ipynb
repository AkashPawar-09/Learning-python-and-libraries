```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = [2, 3, 1,2, 4 ,3, 8 ,3, 4, 5, 5, 5, 6,7,7]
plt.hist(data, bins=10, color='pink', edgecolor='r', label=None, density=False)
plt.show()
```

```python
data = np.random.randn(1000)

plt.hist(data, bins=20, color='skyblue', edgecolor='black')

plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Basic Histogram')
plt.show()

# np.random.randn(1000) → Generates 1000 random numbers.
# bins=30 → Divides the data range into 30 intervals.
# Frequency → Number of values falling into each interval.
```

```python
x = np.random.normal(170, 10, 250)
# np.random.normal(mean, standard_deviation, number_of_values)
plt.hist(x)
```

| Parameter              | Value | Meaning                       |
| ---------------------- | ----- | ----------------------------- |
| Mean (μ)               | 170   | Average value                 |
| Standard deviation (σ) | 10    | How spread out the values are |
| Size                   | 250   | Generates 250 random values   |


```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

data = np.random.randn(1000)
ax = sns.histplot(data, bins=30, kde=True, color='cyan',edgecolor='grey',line_kws={'linewidth': 2.5,'linestyle': '--'})
ax.lines[-1].set_color('hotpink')
plt.xlabel('Values')
plt.ylabel('Density')
plt.title('Customized Histogram with Density Plot')
plt.show()
```
