## Trignometric functions plotting 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2 * np.pi, 100)
# np.linspace(start, stop, num) generates num equally spaced values between start and stop.y_sin = np.sin(x)

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y_sin, label="sin(x)")
plt.plot(x, y_cos, label="cos(x)")
plt.xlabel("X values")
plt.ylabel("Function value")
plt.title("Trigonometric Functions")
plt.legend()
plt.show()
```

## All the functions are taken from AI to just show different graphs in Math

```python
n = np.arange(1, 101)

o1 = np.ones(len(n))
ologn = np.log2(n)
on = n
onlogn = n * np.log2(n)
on2 = n**2

plt.plot(n, o1, label='O(1)')
plt.plot(n, ologn, label='O(log n)')
plt.plot(n, on, label='O(n)')
plt.plot(n, onlogn, label='O(n log n)')
plt.plot(n, on2, label='O(n²)')

plt.xlabel('Input Size (n)')
plt.ylabel('Operations')
plt.title('Time Complexity Comparison')
plt.legend()
plt.grid()
plt.show()
```

```python
x = np.linspace(1, 10, 100)

y1 = 2**x
y2 = np.log2(x)

plt.plot(x, y1, label='Exponential')
plt.plot(x, y2, label='Logarithmic')

plt.xlabel('Input')
plt.ylabel('Growth')
plt.title('Exponential vs Logarithmic')
plt.legend()
plt.grid()
plt.show()
```

```python
x = np.linspace(-10, 10, 200)

y = 1 / (1 + np.exp(-x))

plt.plot(x, y, label='Sigmoid')
plt.xlabel('Input')
plt.ylabel('Output')
plt.title('Sigmoid Function')
plt.legend()
plt.grid()
plt.show()
```

```python
x = np.linspace(-10, 10, 200)

y = np.maximum(0, x)

plt.plot(x, y, label='ReLU', c='red')
plt.xlabel('Input')
plt.ylabel('Output')
plt.title('ReLU Activation Function')
plt.legend()
plt.grid()
plt.show()
```

```python
epochs = np.arange(1, 101)

loss = np.exp(-0.05 * epochs) + 0.05

plt.plot(epochs, loss, label='Training Loss', c='red')

plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Loss vs Epochs')
plt.legend()
plt.grid()
plt.show()
```

```python
t = np.arange(1, 101)

lr = 0.1 / (1 + 0.05*t)

plt.plot(t, lr, label='Learning Rate')

plt.xlabel('Iterations')
plt.ylabel('Learning Rate')
plt.title('Learning Rate Decay')
plt.legend()
plt.grid()
plt.show()
```
