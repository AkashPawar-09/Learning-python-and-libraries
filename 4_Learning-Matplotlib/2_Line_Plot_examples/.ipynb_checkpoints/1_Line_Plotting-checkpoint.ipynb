````python
price = [48000,54000,57000,50000,45000,40000,49000]
year = [2015,2016,2017,2018,2019,2020,2021]
for xi, yi in zip(year, price):
    plt.annotate(f'({xi}, {yi})',
                 (xi, yi),
                 textcoords="offset points",
                 xytext=(0, 10),
                 fontsize='x-small',
                 c = 'dodgerblue' ,
                 ha='center')

plt.title('Line Chart with Annotations')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.tick_params(axis='y', colors='crimson')
plt.tick_params(axis='x', colors='green')
plt.grid(True)
plt.plot(year,price)
plt.show()
````

````python
x = np.array([1, 2, 3, 4])
y = x * 1.3

x1 = [2, 4, 6, 8]
y1 = x * 0.7

plt.plot(x, y, label='first line', c='r', linestyle='--',marker='*', ms=10 , lw = 1 , mec = 'gold' , mfc = 'gold')
plt.plot(x1, y1, label='Second Line', c='b',linestyle='-', marker='v', ms=7 , lw = 1 , mec = 'red', mfc = 'red')
plt.xlabel("X-axis data")
plt.ylabel("Y-axis data")
plt.title("Multiple Plots on Same Axis")
plt.legend()
plt.show()
````

### Understanding `fill_between()`
Syntax:
```
plt.fill_between(x, y1, y2, color='green', alpha=0.4)
```
| Parameter | Meaning                |
| --------- | ---------------------- |
| `x`       | Common X-axis values   |
| `y1`      | First line's Y values  |
| `y2`      | Second line's Y values |
| `color`   | Fill area color        |
| `alpha`   | Transparency (0 to 1)  |

````python
x = np.array([1, 2, 3, 4])
y = x * 1.3

x1 = [2, 4, 6, 8]
y1 = x * 0.7

plt.plot(x, y)
plt.plot(x1, y1)
plt.fill_between(x, y, y1, color='green', alpha=0.4)

plt.xlabel("X-axis data")
plt.ylabel("Y-axis data")
plt.title("Filled Area Between Two Lines")
plt.legend()
plt.show()
````

### Important: Both lines must use the same X coordinates for this simple fill_between()

````python
x = np.array([1, 2, 3, 4])
y1 = x * 1.3
y2 = x * 0.7

plt.plot(x, y1, label='y1 = 1.3x')
plt.plot(x, y2, label='y2 = 0.7x')

plt.fill_between(x, y1, y2, color='green', alpha=0.4)

plt.xlabel("X-axis data")
plt.ylabel("Y-axis data")
plt.title("Filled Area Between Two Lines")
plt.legend()
plt.savefig("line_plot.png")
plt.show()
````
