```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [ 1, 2, 4, 9, 12, 16]
y = [ 4, 3, 10, 12, 17, 20]
plt.plot(x, y)        # plot x and y using default line style and color
```

```python
plt.plot(y)           # plot y using x as index array 0..N-1
# y = [ 4, 3, 10, 12, 17, 20]
```

```python
plt.plot(x, y, 'o')  # plot x and y using circle markers
```

```python
plt.plot(x, y, 'bo')  # plot x and y using blue circle markers
```

```python
plt.plot(y, '+')     # ditto, but with plusses
```

```python
plt.plot(y, 'r+')     # ditto, but with red plusses
```

```python
# If x and/or y are 2D arrays, a separate data set will be drawn for every column. If both x and y are 2D, they must have the same shape. 
# If only one of them is 2D with shape (N, m) the other must have length N and will be used for every data set m.
x = [1, 2, 3]
y = np.array([[1, 2], [3, 4], [5, 6]])
plt.plot(x, y)
```

```python
for col in range(y.shape[1]):
    plt.plot(x, y[:, col])
```

```python
plt.plot(x, y, 'go--', linewidth=2, markersize=15)      # 'go--' ----> green , circle , --
```

```python
plt.plot(x, y, color='green', marker='o', linestyle='dashed',
     linewidth=2, markersize=15)
```

```python
plt.plot([1, 2, 3], [1, 2, 3], 'go-', label='line 1.5', linewidth=3)     # green , circle marker , linestyle - (dashed)
```

```python
plt.plot([1,2,3],[4,5,7] , color=(0.1, 0.5, 0.8))
# Matplotlib expects RGB/RGBA values to be tuples or lists of floats ranging between 0.0 and 1.0.
# By putting quotes around it, you passed a string text "(0.1, 0.2, 0.5)", which Matplotlib tries and fails to look up in its named color database (like 'red', 'blue', etc.).
```

```python
plt.plot([1,2,3],[4,5,7] , color='#0f0f0f')
# Case-insensitive hex RGB or RGBA string
```

```python
plt.plot([1,2,3],[6,7,10] , color='0')     # '0' as black
# String representation of float value in closed interval [0, 1] for grayscale values.
```

```python
plt.plot([1,2,3],[6,7,10] , color='1')     # '1' as white
```

```python
fig, ax = plt.subplots()
ax.plot([1, 2, 3], [6, 7, 10], color='1')  # White line

ax.set_facecolor('0.8')  # light gray background
plt.show()
```

```python
plt.plot([1,2,3],[6,7,10] , color='w')     # 'w' as white
plt.gca().set_facecolor('0.8')              # Get current axes and set light gray background
```

### Bold, high-contrast colors make your data points, labels, and text pop immediately against a white or light background.
---
### 🔴 High-Visibility Primary & Vibrant Colors
### red / crimson / firebrick (Extremely sharp and demanding attention) • blue / navy / dodgerblue (Deep, highly legible, and professional) • green / forestgreen / lime (Bright and distinct)
---
### 🦩 Bright Pinks & Purples
### hotpink / deeppink / magenta (Intense, glowing vibrancy) • purple / darkviolet / indigo (Rich tone with excellent text legibility)
---
### 🍊 Neon Oranges & Golds
### darkorange / orange (Warm, bright, and very punchy) • gold (Highly visible, though best paired with black text borders)
---
### 🐬 Sharp Cyans & Teals
### teal / darkcyan (Excellent dark-yet-vibrant alternative to standard blue) • deepskyblue / cyan (Electric pop, best used for lines rather than text)

```python
plt.plot([1,2,3],[6,7,10] , color='b')     # 'b' as blue
"""
'b' as blue
'g' as green
'r' as red
'c' as cyan
'm' as magenta
'y' as yellow
'k' as black
'w' as white
"""
```

```python
plt.plot([1,2,3],[6,7,10] , color='green', marker='o', linestyle='-',
     linewidth=2, markersize=15 )
```
