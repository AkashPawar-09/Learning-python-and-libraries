```python
import numpy as np
```

```python
## file related operations 
```

```python
file_data = np.genfromtxt('numbers.txt' ,delimiter = ',' )
print(file_data)
```

```python
file_data.astype('int32')
```

```python
## Boolean Masking and Advanced Indexing 
```

```python
file_data > 30     # bool value for numbers greater than 30 in matrix file_data
```

```python
file_data[file_data > 30]    # numbers which are greater than 30
```

```python
np.genfromtxt("numbers.txt", delimiter=",", usemask=True)
```

```python
np.genfromtxt("numbers.txt", delimiter=",", usemask=True).astype('int32')
```

```python
file_2 = np.genfromtxt('two_D_matrix.txt' ,delimiter = ',' )
print(file_2)
```

```python
file_2.astype('int32')
```

```python
np.any(file_2 > 30 , axis=0)    # any number is less than 30 in cols
```

```python
np.any(file_2 > 43 , axis=1)    # any number is greater than 42 in rows
```

```python
((file_2 > 5) & (file_2 < 45 ))
```

```python
(~((file_2 > 5) & (file_2 < 45 )))
```
