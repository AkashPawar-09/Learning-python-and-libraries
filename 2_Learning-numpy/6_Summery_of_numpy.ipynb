## 7 BOOLEAN MASKING / FILTERING

```python
data = np.array([5, 12, 7, 20, 3, 15])
print(data)
```

```python
mask = data > 10
print("masking for data which is smaller than 10 : ", mask)
# Comparison gives True/False array.
```

```python
print("mark numbers high and low based on value (low <= 10 < high) :", np.where(data > 10, "high", "low"))
# np.where(condition, if_true, if_false) -> builds new array. Great for labels.
```

## 8 MISSING VALUES (NaN) - DATA CLEANING

```python
m = np.array([1.0, np.nan, 3.0, np.nan, 5.0])
print(m)
# np.nan = missing value (works only with float arrays).
```

```python
print("value missing or not :", np.isnan(m))
# True where value is missing.
```

```python
print("Count missing values :", np.sum(np.isnan(m)))
# Count of missing values -> 2.
```

```python
print("print element which is nan in m:", m[np.isnan(m)])
print("print element which is not nan in m::", m[~np.isnan(m)])
# '~' flips True/False -> keeps only non-missing values.
```

```python
print("ignoring nan ----> mean (3/3) and sum (1+3+5) is :", np.nanmean(m), np.nansum(m))
# nan-versions ignore NaN. Normal m.mean() would return nan.
```

```python
m2 = np.where(np.isnan(m), np.nanmean(m), m)
print("Replacing all nan elements by mean of original array:", m2)
# Replace NaN with mean (very common fill strategy).
```

```python
print("Replace nan by desired number : ", np.nan_to_num(m, nan=0))
```

## 14 OUTLIERS

#### This section demonstrates how to detect outliers in a dataset using the Interquartile Range (IQR) method, which is a widely used statistical technique.

```python
v = np.array([10, 12, 11, 13, 12, 100])
print(v)

# Notice that 10, 12, 11, 13, 12 are all closely grouped together, while 100 stands out as extremely large.
```

```python
q1, q3 = np.percentile(v, [25, 75])
print("q1 :",q1)
print("q3 :",q3)
iqr = q3 - q1      # Calculating the IQR (Interquartile Range)
print(iqr)         # IQR = spread of the middle 50% of data.

# IQR is the distance between the 75th and 25th percentiles (12.75 - 11.25 = 1.5)
# Q1 (25th Percentile): The value below which 25% of the data falls. The output shows 11.25.
# Q3 (75th Percentile): The value below which 75% of the data falls. The output shows 12.75.
```

```python
low  = q1 - 1.5 * iqr   # Low Boundary (9.0): Any value lower than 9.0 is a low outlier.
high = q3 + 1.5 * iqr   # High Boundary (15.0): Any value higher than 15.0 is a high outlier.
print("Low :",low,"AND High :",high)     # Our acceptable range for normal data is 9.0 to 15.0.
```

#### Finding the Outliers

```python
print("outliers:", v[(v < low) | (v > high)])   

# It looks for values that are either strictly less than 9.0 or strictly greater than 15.0.
# Why use it: This is how you mathematically identify and isolate the anomalies in your dataset.
```

```python
print("cleaned:", v[(v >= low) & (v <= high)])    

# It only preserves elements that fall safely within the acceptable range—meaning 
# they must be greater than or equal to 9.0 and less than or equal to 15.0
# Why use it: This is the most common data cleaning step when you want to completely drop extreme anomalies 
# so they do not skew your machine learning models or statistical averages.
```

```python
print("14c clipped:", np.clip(v, low, high))

# Instead of throwing data away, np.clip() modifies the extreme values. Any number smaller than low is automatically bumped up to 9.0, 
# and any number larger than high is capped down to 15.0. All normal values in between stay exactly the same.

# Output: [10., 12., 11., 13., 12., 15.] (Note: 100 was changed to 15.0)
# This is useful when you cannot afford to lose data rows but you still need to minimize the destructive impact of an extreme value. 
```
