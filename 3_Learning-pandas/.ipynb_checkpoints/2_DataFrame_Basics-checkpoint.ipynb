```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Amit", "Riya", "Sam", "Neha", "Raj"],              # stored as key value pairs 
    "Age": [21, 22, 20, 23, 21],
    "City": ["Pune", "Mumbai", "Delhi", "Pune", "Nagpur"],
    "Marks": [85, 90, 78, 92, 88]
}
```

```python
df = pd.DataFrame(data)             # creating a data frame 
print(df)
```

```python
# The index 0 to 4 is created automatically.
```

```python
rows = [["Amit", 21], ["Riya", 22], ["Sam", 20], ["Neha" ,23]]       # creating a new 2D array (rows)
```

```python
df_list = pd.DataFrame(rows, columns=["Name", "Age"])      # name and age cols are created from rows 
print(df_list)
```

```python
records = [                                           # Creating a new dictionary
    {"Name": "Amit", "Age": 21},                      # 2 cols    (1 missing)
    {"Name": "Riya", "Age": 22, "City": "Mumbai"}     # 3 cols    (all 3 cols)
]
df_rec = pd.DataFrame(records)                        # converting records dict to dataframe
print(df_rec)                                         # NaN = Not a Number (meaning missing)
```

```python
# np.arange(1, 10) creates numbers 1 to 9.
```

```python
arr = np.arange(1, 10).reshape(3, 3)            # creating 1 to 9 nums and then shaping them into 3*3
print(arr)
```

```python
df_np = pd.DataFrame(arr, columns=["A", "B", "C"], index=["r1", "r2", "r3"])     # name the cols and then index(rows)
print(df_np)
```

```python
s1 = pd.Series([85, 90, 78], index=["Amit", "Riya", "Sam"])     # creating new series with indices 
s2 = pd.Series([21, 22, 20], index=["Amit", "Riya", "Sam"])     # creating new series with same indices 
df_s = pd.DataFrame({"Marks": s1, "Age": s2})                   # Creating dataframe using s1 and s2 and naming cols as marks and age
print(df_s)
```

## Basic column operations on a DataFrame

```python
print(df["Marks"] + 5)       # A column is a Series, so vectorized math works exactly as before.
```

```python
# df.columns is a pandas Index containing the column names. 
# .tolist() converts it into a regular Python list.
```

```python
df["Percentage"] = df["Marks"] / 100 * 100       # Add a new column
print(df.head())
```

```python
df["Passed"] = df["Marks"] > 80                      # Add a True/False column
print(df[["Name", "Marks", "Passed"]])
```

```python
df = df.drop(columns=["Percentage", "Passed"])        # Delete a columns named percentage and 
print(df.columns.tolist())                            # new list of cols 
```

```python
# df["Marks"].mean() works on one column and gives a single number.
# df.mean(numeric_only=True) works on all numeric columns and gives one result per column. 
# numeric_only=True skips the text columns, otherwise pandas may raise an error.
```

```python
print(df["Marks"].mean())
```

```python
print(df["Age"].max())
```

```python
print(df.mean(numeric_only=True))        # for cols having numeric values give their mean
```

```python
print(df)
```

```python
print(df.T)                             # T swaps rows and columns, exactly like NumPy’s .T   (T -> Transpose)
```

```python
# copy() creates an independent DataFrame, so changing df_copy does not change df. loc[0, "Marks"] means row label 0, column “Marks”.
```

```python
df_copy = df.copy()
```

```python
df_copy.loc[0, "Marks"] = 0
```

```python
print(df.loc[0, "Marks"])
```

```python
print(df_copy)
```

```python
df_copy.loc[:, "Marks"] = 0
```
