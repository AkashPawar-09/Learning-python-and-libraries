```python
# iloc means integer location. It ignores labels and uses only positions, like NumPy indexing.
```

```python
print(words.iloc[0])
```

```python
print(words.iloc[[0, 2, 4]])
```

```python
# 2:5 means positions 2, 3, 4. The end is excluded, as in normal Python.
# -2: means the last two rows. Negative positions count from the end./
```

```python
print(words.iloc[2:5])
print(words.iloc[-2:])
```

```python
# loc[rows, columns] takes labels on both sides, so you can write column names instead of positions.
```

```python
print(words.loc[3, "Word"])
```

```python
print(words.loc[[0, 3], ["Word", "Value"]])
```

```python
print(words.loc[2:4])
```

```python
print(words.loc[2:4, "Word":"Char Count"])
```

```python
# The same-looking slice returns 2 rows with iloc and 3 rows with loc. This single fact causes many bugs.
```

```python
print(words.iloc[2:4].shape)
print(words.loc[2:4].shape)
```

## Single cells: at and iat

```python
print(words.iloc[3])
```

```python
# at[row label, column name] gets one cell by label.
# iat[row position, column position] gets one cell by position.
# They are faster than loc and iloc when you only need one value.
```

```python
print(words.at[3, "Word"])
```

```python
print(words.iat[3, 0])
print(words.iat[3, 1])
print(words.iat[3, 2])
```

```python
# & means AND: 2-letter words with Value 40 or more. There are 5.
```

```python
print(len(words[(words["Char Count"] == 2) & (words["Value"] >= 40)]))
```

```python
print(words[(words["Char Count"] == 2) & (words["Value"] >= 40)])
```

```python
# number whose Char Count is greater than 25 OR whose value is geater than 315
```

```python
print(len(words[(words["Char Count"] > 25 ) | (words["Value"] > 315)]))     # total numbers
```

```python
print(words[(words["Char Count"] > 25 ) | (words["Value"] > 315)])          # actual values
```

```python
print(len(words[(words["Char Count"] <= 20)]))
```

```python
print(len(words[~(words["Char Count"] <= 20)]))
```

```python
print(len(words[(words["Char Count"] > 20)]))
```

```python
print(len(words[(words["Char Count"] <= 20)]) + (len(words[~(words["Char Count"] <= 20)])))
```

## isin and between

```python
print(words[words["Char Count"].isin([27, 28])])   
# wodrs having Char count 27 , 28
```

```python
print(words[words["Word"].isin(["python", "pandas", "numpy"])])
```

```python
print(words[words["Word"].isin(["numpy"])])
```

```python
print(words[words["Value"].between(300, 310)].shape)  # rows from 300 to 309
```
