### Files without a header, 

```python
import io
```

```python
print(pd.read_csv(io.StringIO(text), header=None))
```

### Files with a different separator

```python
# CSV means comma-separated, but some files use ;, a tab \t, or |. sep=";" tells pandas the real separator. For tab-separated files use sep="\t".
```

```python
text2 = "letter;num\na;1\nb;2"
print(text2)
```

```python
print(pd.read_csv(io.StringIO(text2), sep=";"))
```

```python
text3 = "item,price\npen,10\nbook,-"
```

```python
print(text3)
```

```python
# na_values=["-"] treats the - symbol as a missing value (NaN).
# Without it, the price column would be read as text, because - is not a number.
# The price column becomes float because NaN is a float.
```

```python
print(pd.read_csv(io.StringIO(text3), na_values=["-"]))              # Custom missing markers: na_values
```

```python
# chunksize=50000 does not load everything at once. It returns pieces of 50,000 rows each.
# The for loop gives you one DataFrame (chunk) per piece.
# The last chunk is smaller because 172,823 is not a multiple of 50,000.
# Use this when a file is too big for your memory.
```

```python
for chunk in pd.read_csv("words.csv", chunksize=50000):
    print(chunk.shape)      # shape -> rows , cols
```

```python
for chunk in pd.read_csv("words.csv", chunksize=100000):
    print(chunk.shape)      # shape -> rows , cols
```

## Writing a CSV File

```python
subset = words.head(3)                           # subset is new group of first 3 words
subset.to_csv("small.csv", index=False)          # make this group as small.csv 
```

```python
# to_csv("small.csv") saves the DataFrame to a new file.
# index=False stops pandas from writing the 0, 1, 2 index as an extra column.
```

```python
print(open("small.csv").read())                  # open("small.csv").read() reads the file back as plain text.
```

```python
subset.to_csv("with_index.csv")
print(pd.read_csv("with_index.csv").head(2))
```

```python
subset.to_csv("selected.csv", index=False)
print(pd.read_csv("selected.csv").head(2))
```

```python
subset.to_csv("selected.csv", index=False, columns=["Word", "Value"], sep="|")
print(open("selected.csv").read())
```

```python
# columns=[...] writes only those columns.
# sep="|" uses a different separator.
```

```python
# to_excel(...) saves an .xlsx file. sheet_name="Words" names the sheet.
# read_excel(..., sheet_name="Words") reads that sheet. By default it reads the first sheet.
# Excel files are slower than CSV, so use CSV for large data.
```

```python
subset.to_excel("small.xlsx", index=False, sheet_name="Words")
```

```python
back = pd.read_excel("small.xlsx", sheet_name="Words")
print(back)
```

```python
with pd.ExcelWriter("multi.xlsx") as writer:
    words.head(3).to_excel(writer, sheet_name="First", index=False)
    words.tail(3).to_excel(writer, sheet_name="Last", index=False)
```

```python
# pd.ExcelWriter("multi.xlsx") opens one Excel file for writing.
# with ... as writer makes sure the file is saved and closed properly when the block ends.
# Each to_excel(writer, ...) adds one sheet to the same file.
```

```python
sheets = pd.read_excel("multi.xlsx", sheet_name=None)
print(list(sheets.keys()))
```

```python
# sheet_name=None reads all sheets and returns a dictionary of {sheet name: DataFrame}. keys() lists the sheet names.
```
