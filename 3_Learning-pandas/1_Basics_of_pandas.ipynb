## Basics of pandas

```python
import pandas as pd
import numpy as np

print(pd.__version__)
```

```python
marks = pd.Series([85, 90, 78, 92, 88])
print(marks)
```

```python
marks = pd.Series(
    [85, 90, 78, 92, 88],
    index=["Amit", "Riya", "Sam", "Neha", "Raj"],
    name="Maths"
)
print(marks)
```

```python
# .values gives the underlying NumPy array.
# .index gives the labels.
# .dtype gives the data type of the values.
# .shape gives a tuple with the number of elements.
# .size gives the total count of elements.
# .ndim gives the number of dimensions (always 1 for a Series).
# .name gives the Series name.
```

```python
print(marks.values)
print(marks.index)
print(marks.dtype)
print(marks.shape)
print(marks.size)
print(marks.ndim)
print(marks.name)
```

```python
a = pd.Series([10, 20, 30], index=["x", "y", "z"])
b = pd.Series([1, 2, 3], index=["y", "z", "w"])
print(a + b)
```

```python
# Pandas matches values by label, not by position. y and z exist in both, so they are added. x and w exist in only one, 
# so the result is NaN (Not a Number, meaning missing). The dtype becomes float64 because NaN is a float.

# x + ? = NaN (Not a Number)
# y + y = 20+1 = 21
# z + z = 30+2 = 32
# w + ? = NaN (Not a Number)
```

## Unique values and counts

```python
colors = pd.Series(["red", "blue", "red", "green", "blue", "red"])
```

```python
print(colors)
```

```python
# unique() returns the distinct values as an array.
# nunique() returns how many distinct values there are.
# value_counts() counts how many times each value appears, sorted from most to least frequent.
```

```python
print(colors.unique())
```

```python
print(colors.nunique())
```

```python
print(colors.value_counts())
```

## Missing values (basic)

```python
# np.nan represents a missing value.
# isna() returns True where a value is missing.
# sum() and mean() skip NaN by default. The mean is 90/3 = 30, not 90/5.
# fillna(0) replaces every missing value with 0.
# dropna() removes the missing entries.
```

```python
data = pd.Series([10, np.nan, 30, np.nan, 50])
```

```python
print(data)
```

```python
print(data.isna())
```

```python
print(data.sum())
```

```python
print(data.mean())
```

```python
print(data.fillna(0))
```

```python
print(data.dropna())
```
