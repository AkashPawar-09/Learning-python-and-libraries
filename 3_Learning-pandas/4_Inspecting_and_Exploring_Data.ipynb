```python
import pandas as pd

words = pd.read_csv("words.csv")
```

```python
print(words.shape)
```

```python
print(words.head())
```

```python
print(words.tail())
```

```python
# sample(3) picks 3 random rows. Because the file is sorted, 
# head() only shows words starting with “a”, so a sample gives a fairer picture.
# random_state=1 fixes the randomness, so you get the same rows every time. 
# Without it, the rows change on each run.
```

```python
print(words.sample(3, random_state=1))      # as here state is fixed so everytime same randoms...
```

```python
words.info()
```

```python
print(words.describe())
```

```python
# include="all" adds the text column too.
# unique is the number of distinct words, 172,821.
# top is the most frequent value and freq is how many times it appears. 
# Every word appears once, so top is just the first one.
# NaN appears where a statistic makes no sense for that column type.
```

```python
print(words.describe(include="all"))
```

```python
# isna() gives a True/False table with True where a value is missing.
# .sum() counts the True values in each column.
```

```python
print(words.isna().sum())
```

```python
print(words.isna())
```

```python
print(words["Word"])
```

```python
# The surprise. Row 96714 has 3 letters and a Value of 29, and row 100730 has 4 letters and a Value of 59. 
# These are real words, not missing data. 
# Letter values give 29 = n+a+n and 59 = n+u+l+l. The words are nan and null.
# Pandas treats the text nan, null, NA, None and a few others as missing by default, 
# so it converted them to NaN while reading
```

```python
print(words[words["Word"].isna()])
```

```python
# keep_default_na=False stops pandas from converting those texts to NaN.
# Now nothing is missing and the two words are back.
```

```python
fixed = pd.read_csv("words.csv", keep_default_na=False)
```

```python
print(fixed.isna().sum())
```

```python
print(fixed.loc[[96714, 100730]])
```

```python
# A second .sum() adds the per-column counts into one total, 2.
# isna().mean() gives the fraction of missing values, because True counts as 1. It is about 0.0012% of rows.
# 1.157e-05 is scientific notation, meaning 0.00001157.
```

```python
print(words.isna().sum().sum())
```

```python
print(words["Word"].isna().mean())
```

```python
# dropna() removes rows that contain any missing value, leaving 172,821 rows.
```

```python
print(words.dropna().shape)
```

```python
# count() gives the non-missing values per column.
# nunique() gives the number of distinct values per column, ignoring NaN.
# Char Count has only 26 distinct lengths and Value has 303 distinct values.
```

```python
print(words.count())
```

```python
print(words.nunique())
```

```python
print(words.head(3))
```

```python
print(words["Char Count"].value_counts().head(3))
```

```python
print(words["Char Count"].head(3).value_counts())
```

```python
# sort_index() sorts by the length itself instead of by frequency, so you can read the distribution in order.
# No word has 26 letters, so the length 26 is missing from the list.
# The counts rise to a peak at 8 letters and then fall, which is the shape of a bell curve.
```

```python
print(words["Char Count"].value_counts().sort_index())
```

```python
print(words[words["Char Count"] == 25])
```

```python
print(words[words["Char Count"] == 27])
```

```python
print(words[words["Char Count"] == 28])
```

```python
# duplicated() marks a row True when it is an exact copy of an earlier row. The sum is 0, so there are no duplicate rows.
# is_unique is False for the Word column because the two NaN values count as the same value, so they are not unique.
```

```python
# subset=["Value"] checks duplicates using only that column. 172,520 rows repeat a Value that appeared earlier, 
# which makes sense: only 303 distinct values exist across 172,823 words.
```

```python
print(words.duplicated(subset=["Value"]).sum())
```

```python
# nsmallest(3, "Value") returns the 3 rows with the lowest Value.
```

```python
print(words.nsmallest(4, "Value"))
```

```python
# nlargest(3, "Value") returns the 3 rows with the highest Value.
```

```python
print(words.nlargest(4, "Value"))
```

```python
# idxmax() returns the index label of the largest value, here 51072.
```

```python
print(words["Char Count"].idxmax())
```

```python
# loc[51072] then fetches that whole row. It is the longest word in the file, with 28 letters.
```

```python
print(words.loc[words["Char Count"].idxmax()])
```

```python
# agg([...]) runs several statistics at once from a list of names.
```

```python
print(words["Value"].agg(["min", "max", "mean", "median"]))
```

```python
# quantile([0.1, 0.9]) gives the value below which 10% and 90% of the data fall. 
# So 80% of the words have a Value between 61 and 160.
```

```python
print(words["Value"].quantile([0.1, 0.9]))
```

```python
print(words["Value"].quantile([0.2, 0.7]))
```

```python
print(words["Value"].quantile([0.4, 0.5]))
```

```python
print(words["Value"].quantile([0.49, 0.50]))
```

```python
# mode() returns the most frequent value, here 8. It returns a Series because there can be ties.
```

```python
print(words["Char Count"].mode())
```

## Memory usage

```python
# memory_usage() gives the bytes used by each column and the index.
# Each column holds 172,823 values at 8 bytes each, which is 1,382,584 bytes.
```

```python
print(words.memory_usage())
```

```python
# deep=True also counts the real size of the text. The full table then takes about 12.8 MB, much more than the 4.0 MB that info() showed, 
# because info() does not count the text contents by default.
```

```python
print(words.memory_usage(deep=True).sum())
```
