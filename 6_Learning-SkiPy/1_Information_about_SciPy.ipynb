## Functions and Modules

### 1. Optimization (`scipy.optimize`)

- **minimize:** Finds the minimum of scalar or multivariable functions using various algorithms.
- **root:** Solves equations or systems of nonlinear equations to find roots.
- **curve_fit:** Performs least squares fitting of a function to data.
- **linprog:** Solves linear programming problems.

### 2. Integration (`scipy.integrate`)

- **quad:** Performs adaptive quadrature for definite integrals.
- **dblquad and tplquad:** Compute double and triple integrals.
- **odeint:** Solves ordinary differential equations (ODEs).
- **solve_ivp:** Solves initial value problems of ODEs using various methods.

### 3. Interpolation (`scipy.interpolate`)

- **interp1d:** Performs one-dimensional interpolation with different methods.
- **griddata:** Interpolates scattered data points on a grid.
- **BarycentricInterpolator:** Provides efficient polynomial interpolation.

### 4. Linear Algebra (`scipy.linalg`)

- **solve:** Solves linear systems of equations.
- **inv:** Computes the inverse of a matrix.
- **eigh:** Computes eigenvalues and eigenvectors of Hermitian matrices.
- **svd:** Performs Singular Value Decomposition.

### 5. Statistics (`scipy.stats`)

- **norm:** Provides functions for the normal (Gaussian) distribution.
- **ttest_ind:** Performs t-tests on two independent samples.
- **pearsonr:** Computes the Pearson correlation coefficient.
- **describe:** Summarizes data using descriptive statistics.

### 6. Signal Processing (`scipy.signal`)

- **convolve:** Computes the convolution of two signals.
- **butter:** Designs Butterworth filters.
- **find_peaks:** Detects peaks in data.
- **spectrogram:** Computes the spectrogram of a signal.

### 7. Fourier Transforms (`scipy.fft`)

- **fft:** Computes the one-dimensional Discrete Fourier Transform.
- **ifft:** Computes the Inverse Fourier Transform.
- **fft2:** Computes the two-dimensional Fourier Transform.

### 8. Sparse Matrices (`scipy.sparse`)

- **csr_matrix:** Creates a Compressed Sparse Row matrix.
- **csc_matrix:** Creates a Compressed Sparse Column matrix.
- **spsolve:** Solves sparse linear systems efficiently.

## Applications of SciPy

1. **Scientific Research and Engineering:** SciPy is widely used to model physical systems, solve differential equations and perform numerical simulations in fields such as physics, chemistry, biology and engineering.

2. **Data Analysis and Statistics:** SciPy provides statistical functions for hypothesis testing, descriptive statistics and distribution fitting. Data scientists use it to process datasets, perform statistical tests and generate probabilistic models.

3. **Signal and Image Processing:** SciPy provides tools for signal filtering, Fourier transforms and spectral analysis of audio, radar and biomedical signals. Modules such as `scipy.ndimage` support image filtering, transformations and image measurements.

4. **Optimization Problems:** SciPy's optimization routines solve linear and nonlinear optimization problems in machine learning, finance and operations research. They are used for parameter tuning, resource allocation and risk management.

5. **Machine Learning and AI:** SciPy acts as a foundational library for machine learning frameworks by providing functions for data preprocessing, feature extraction and mathematical operations used in various algorithms.

# SciPy - Stats

The `scipy.stats` is a SciPy sub-package mainly used for probability distributions and statistical operations. It provides a wide range of probability and statistical functions.

There are three classes:

| Class | Description |
|---|---|
| `rv_continuous` | Used for continuous random variables. We can create specialized distribution subclasses and instances. |
| `rv_discrete` | Used for discrete random variables. We can create specialized distribution subclasses and instances. |
| `rv_histogram` | Used to generate probability distributions based on histograms. |

## Continuous Random Variables

A continuous random variable is a variable that can take any value within a given range.

- **Mean (`loc`):** Defines the location or average of the distribution.
- **Standard Deviation (`scale`):** Determines the spread of the distribution.

Using the `rv_continuous` class, we can create specialized distribution subclasses and instances.

There is a method called `norm`, which is based on `rv_continuous` and represents the Normal (Gaussian) distribution.

The `norm` function provides methods such as CDF (Cumulative Distribution Function), PDF (Probability Density Function) and PPF (Percent Point Function).
