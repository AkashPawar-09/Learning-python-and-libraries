# 1. Optimization (scipy.optimize)

---
## 1. minimize()
## Finds the minimum value of a single-variable or multivariable function.

```python
from scipy.optimize import minimize

def fun(x):
    return x**2 + 5*x + 6

result = minimize(fun, x0=0)

print(result.x)
print(result.fun)
print(result.success)
```

### Explanation:
- fun: Function whose minimum we want to find.
- x0: Initial guess for the starting point.
- result.x: Value of x at the minimum.
- result.fun: Minimum function value.
- result.success: Tells whether optimization was successful.

---
# 2. root()
## Finds the root of an equation where the function value becomes zero.

```python
from scipy.optimize import root

def fun(x):
    return x**2 - 9

result = root(fun, x0=2)

print(result.x)
print(result.success)
```

### Explanation:
- fun: Equation whose root we want to find.
- x0: Initial guess for the root.
- result.x: Calculated root.
- result.success: Indicates whether the solution was successful.

# 3. curve_fit()
## Fits a mathematical function to given data using the least squares method.

```python
import numpy as np
from scipy.optimize import curve_fit

def fun(x, a, b):
    return a*x + b

x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])

popt, pcov = curve_fit(fun, x, y)

print(popt)
print(pcov)
```

### Explanation:
- curve_fit: Fits a mathematical function to data.
- a, b: Unknown parameters of the function.
- popt: Optimized parameter values (a = 2, b = 1).
- pcov: Covariance matrix showing parameter uncertainty.

# 4. linprog()
## Solves Linear Programming problems by minimizing or maximizing a linear objective function subject to constraints.

```python
from scipy.optimize import linprog

c = [2, 3]

A = [[1, 1],
     [2, 1]]

b = [10, 15]

result = linprog(c, A_ub=A, b_ub=b, bounds=(0, None), method="highs")

print(result.x)
print(result.fun)
print(result.success)
```

### Explanation:
- c: Coefficients of the objective function.
- A_ub: Coefficients of inequality constraints.
- b_ub: Upper limits of constraints.
- bounds: Defines the allowed range of variables.
- result.x: Optimal variable values.
- result.fun: Minimum objective function value.

# 5. minimize_scalar()
## Finds the minimum of a single-variable function.

```python
from scipy.optimize import minimize_scalar

def fun(x):
    return (x - 5)**2 + 2

result = minimize_scalar(fun)

print(result.x)
print(result.fun)
```

### Explanation:
- minimize_scalar: Optimizes a function with one variable.
- result.x: Value of x at the minimum.
- result.fun: Minimum function value.

# 6. least_squares()
## Finds the best-fit solution by minimizing the sum of squared residuals.

```python
import numpy as np
from scipy.optimize import least_squares

def fun(x):
    return x**2 - 25

result = least_squares(fun, x0=4)

print(result.x)
print(result.cost)
```

### Explanation:
- least_squares: Minimizes the sum of squared errors.
- x0: Initial guess.
- result.x: Optimized solution.
- result.cost: Half the sum of squared residuals.

# 8. Bounds()
## Defines lower and upper limits for variables during optimization.

```python
from scipy.optimize import minimize, Bounds

def fun(x):
    return (x[0] - 5)**2

bounds = Bounds([0], [3])

result = minimize(fun, x0=[1], bounds=bounds)

print(result.x)
print(result.fun)
```

### Explanation:
- Bounds: Sets limits on optimization variables.
- [0]: Lower bound.
- [3]: Upper bound.
- result.x: Minimum point within the given bounds.
- result.fun: Minimum function value.
