# 3. Barplot

### Barplot is used to compare numerical values across categories. It shows the average value with confidence intervals.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

---

```python
sns.barplot(x="day", y="tip", data=tips)
plt.show()
```

### Conclusion : Day wise Tips...
---

```python
sns.barplot(x="day", y="total_bill", data=tips)
plt.show()
```

### Conclusion : Day wise Profit...
---

```python
sns.barplot(x="smoker", y="total_bill", data=tips)
plt.show()
```

### Conclusion : Smoker spends more...
---

```python
sns.barplot(x="day", y="tip", data=tips, errorbar=None)
plt.show()
```

### Without Error bar
---

```python
sns.barplot(x="day", y="tip", data=tips, errorbar="sd")
plt.show()

# line above each bar is called an Error Bar (Confidence Interval).
# errorbar="sd"	-------> Standard deviation
```

### Standard deviation = How spread out are the actual data values?
