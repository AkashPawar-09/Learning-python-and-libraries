# 1. Lineplot
### Lineplot is used to show the relationship between two continuous variables. It is commonly used for time-series or ordered data.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

#### Example: Visualizing trend over time

```python
fmri = sns.load_dataset("fmri")
```

```python
fmri.head()
```

```python
sns.lineplot(x="timepoint", y="signal", hue="region", data=fmri)
plt.show()
# Explanation: x="timepoint" sets time on x-axis, y="signal" values plotted on y-axis and hue="region" separates lines by region
```

--- 

```python
df = sns.load_dataset("flights")

df.head(10)
```

```python
print(df.shape)
print(df.passengers.mean())
print(df.passengers.max())
print(df.passengers.min())
```

```python
sns.lineplot( data=df, x="year", y="passengers", marker="o" )

plt.title("Yearly Airline Passengers")
plt.xlabel("Year")
plt.ylabel("Total Passengers")
plt.grid(color='gray', linestyle='--', linewidth=0.8)
plt.show()
```

---

```python
flights = sns.load_dataset("flights")
flights.head()
```

```python
sns.lineplot(
    x="month",
    y="passengers",
    hue="year",
    data=flights
)

plt.title("Monthly Airline Passengers by Year")
plt.xlabel("Month")
plt.ylabel("Number of Passengers")
plt.show()
```
