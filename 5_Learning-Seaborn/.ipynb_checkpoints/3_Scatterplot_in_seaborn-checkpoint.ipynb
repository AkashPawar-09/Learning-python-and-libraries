# 2. Scatterplot
### Scatterplot is used to show the relationship between two numerical variables. It helps in identifying patterns or correlations.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
```

#### 💰 Financial Columns

#### • total_bill: The total cost of the meal in U.S. dollars (including taxes).
#### • tip: The gratuity/tip money given to the server in U.S. dollars.

#### 👥 Customer Info

#### • sex: The gender of the person paying the bill (Male or Female).
#### • smoker: Whether anyone in the dining party smoked (Yes or No).
#### • size: The total number of people in the dining party.

#### 📅 Time & Date

#### • day: The day of the week the meal occurred (Thur, Fri, Sat, or Sun).
#### • time: The meal period during which the table was served (Lunch or Dinner).

## Remember:
- Without hue → One common colour for all data points.
- With hue → Different colours for different categories.

```python
sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.show()
```

## Separate colours by day

```python
sns.scatterplot(x="total_bill", y="tip", hue="day", data=tips)
plt.show()
```

## Separate colours by gender

```python
sns.scatterplot(x="total_bill", y="tip", hue="sex", data=tips)
```

## Separate colours by smoker status

```python
sns.scatterplot(x="total_bill", y="tip", hue="smoker", data=tips)
```
