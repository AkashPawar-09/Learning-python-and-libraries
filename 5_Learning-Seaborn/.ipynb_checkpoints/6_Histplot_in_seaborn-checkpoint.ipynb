# 5. Histplot

### Histplot is used to visualize the distribution of a single numerical variable.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
```

```python
tips.head(10)
```

---
### KDE stands for Kernel Density Estimation.
### It adds a smooth curve over the histogram to help visualize the shape of the distribution.

```python
sns.histplot(tips["total_bill"], kde=True)
plt.show()
```

---
## Histogram only

```python
sns.histplot(tips["total_bill"], kde=False)
plt.show()
```

---
## Histogram with KDE

```python
sns.histplot(tips["total_bill"], kde=True)
plt.show()
```

---
### What each parameter does

| Parameter          | Purpose                                     |
| ------------------ | ------------------------------------------- |
| `bins=10`          | Divides the data into 10 intervals          |
| `x="total_bill"`   | Displays total bill on X-axis               |
| `hue="sex"`        | Separates Male and Female using colours     |
| `multiple="stack"` | Stacks both categories on top of each other |
| `kde=True`         | Adds a smooth density curve                 |

```python
sns.histplot( bins=10, data=tips, x="total_bill", hue="sex", multiple="stack",kde=True)
plt.show()
```

```python
sns.histplot( bins=10, data=tips, x="total_bill", hue="sex", multiple="layer",kde=True)
plt.show()
```

```python
sns.histplot( bins=10, data=tips, x="total_bill", hue="sex", multiple="dodge",kde=True)
plt.show()
```

```python
sns.histplot( bins=10, data=tips, x="total_bill", hue="sex", multiple="fill",kde=True)
plt.show()
```

## Seaborn Histplot – Difference between multiple options

| Feature      | `layer`                       | `stack`                                 | `dodge`                            | `fill`                           |
| ------------ | ----------------------------- | --------------------------------------- | ---------------------------------- | -------------------------------- |
| Display      | Histograms overlap each other | Histograms stack on top of each other   | Bars appear side by side           | Bars show proportions            |
| Overlapping  | Yes                           | No                                      | No                                 | No                               |
| Y-axis       | Frequency (count)             | Combined frequency (count)              | Frequency (count)                  | Proportion (0 to 1)              |
| Main purpose | Compare distributions         | Compare total and category counts       | Compare category counts clearly    | Compare category percentages     |
| Best use     | Identify overlapping patterns | See total frequency and contribution    | Compare groups without overlapping | Compare relative proportions     |
| Limitation   | One category may hide another | Individual counts are harder to compare | Takes more horizontal space        | Does not show actual sample size |


---

## 1. `multiple="layer"`

Importance: Helps compare the distribution of multiple categories by drawing histograms on top of one another.

What it gives: You can see where the distributions overlap and where one category has higher frequency than another.

---

## 2. `multiple="stack"`

Importance: Helps understand the total frequency as well as the contribution of each category.

What it gives: Different categories are stacked on top of each other, so the total bar height represents the combined frequency.

Example: Find the total number of male and female customers in each bill range.

---

## 3. `multiple="dodge"`

Importance: Useful when you want to compare different categories clearly without overlapping.

What it gives: Each category gets its own bar next to the other category's bar within the same interval.

Example: Compare male and female customer bill frequencies for each bill range.

---

## 4. `multiple="fill"`

Importance: Helps compare the relative contribution of each category instead of comparing actual counts.

What it gives: Every interval is normalized to 100% (or 1). You can see what percentage of customers belong to each category in that bill range.

Example: Understand what percentage of customers in each bill range are male versus female.

---

## Which one should you use?

| Your requirement               | Best option |
| ------------------------------ | ----------- |
| See overlapping distributions  | `layer`     |
| See combined frequency         | `stack`     |
| Compare actual category counts | `dodge`     |
| Compare category percentages   | `fill`      |

---

Important for Machine Learning and Data Analysis:

- `layer` → Understand distribution overlap.
- `stack` → Understand total frequency and category contribution.
- `dodge` → Compare groups using actual counts.
- `fill` → Compare category proportions independent of the total count.
