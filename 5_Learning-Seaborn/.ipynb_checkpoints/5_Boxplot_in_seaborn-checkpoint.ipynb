# 4. Boxplot

#### Boxplot is used to show the distribution of data based on quartiles. It helps identify spread and outliers.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
```

```python
tips.head(10)
```

```python
sns.boxplot(x="day", y="total_bill", data=tips)
plt.show()
```

### A boxplot has five important components:

| Component   | Meaning                    |
| ----------- | -------------------------- |
| Minimum     | Smallest non-outlier value |
| Q1          | 25th percentile            |
| Median (Q2) | 50th percentile            |
| Q3          | 75th percentile            |
| Maximum     | Largest non-outlier value  |

### IQR (Interquartile Range):

\\[ IQR = Q3-Q1 \\]

### The box represents the middle 50% of the data.

---
### The boxplot uses this rule:

\\[ \text{Upper Limit}=Q3+1.5(IQR) \\]

\\[ \text{Lower Limit}=Q1-1.5(IQR) \\]

### Values outside these limits are considered potential outliers.

---

```python
# Compare tips by gender
sns.boxplot(x="sex", y="tip", data=tips)
```

```python
# Compare total bills by day
sns.boxplot(x="day", y="total_bill", data=tips)
```

```python
# Compare bills by smoker status
sns.boxplot(x="smoker", y="total_bill", data=tips)
```
