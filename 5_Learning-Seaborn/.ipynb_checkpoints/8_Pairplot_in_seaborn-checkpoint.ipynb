# 7. Pairplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris = sns.load_dataset("iris")
```

```python
sns.pairplot(iris, hue="species")
plt.show()
```

## 2. Understanding the 4 × 4 grid

#### Think of the grid like this:

|              | Sepal Length | Sepal Width | Petal Length | Petal Width |
| ------------ | ------------ | ----------- | ------------ | ----------- |
| Sepal Length | Histogram    | Scatter     | Scatter      | Scatter     |
| Sepal Width  | Scatter      | Histogram   | Scatter      | Scatter     |
| Petal Length | Scatter      | Scatter     | Histogram    | Scatter     |
| Petal Width  | Scatter      | Scatter     | Scatter      | Histogram   |

#### There are two types of plots here.


## Explanation: hue adds category distinction.

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

```python
sns.pairplot(tips, hue="sex")
plt.show()
```

---

```python
diamonds = sns.load_dataset("diamonds")
```

```python
diamonds.head()
```

```python
sns.pairplot(
    diamonds.sample(500),
    vars=["carat", "depth", "table", "price"],
    hue="cut"
)
plt.show()
```
