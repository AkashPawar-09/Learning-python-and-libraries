# Linear Algebra

```python
## Matrix Multiplication
```

```python
a = np.ones((2,3))
print(a)
```

```python
b = np.full((3,2),2)
print(b)
```

```python
np.matmul(a,b)
```

## find determinant

```python
c= np.identity(3)
print(c)
```

```python
np.linalg.det(c)
```

```python
a = np.array([[1,2,3],[0, 1, 4],[5, 6, 0]])
print(a)
```

```python
d = np.linalg.inv(a)
print(d)
```

#### Note : if inverse of any matrix is 0 then it gives error

```python
## Dot-Product
```

```python
np.dot(3, 4)
```

```python
np.dot([2j, 3j], [2j, 3j])
```

```python
a = np.array([[1, 0], [0, 1]])
b = np.array([[4, 1], [2, 2]])
print ("Array a :",a)
print("Array b :",b)
print("\nDot Product of a and b : ")
c = np.dot(a, b)
print(c)
```

## Statistics

### find minimum , maximum , sum

```python
stat = np.array([[3,4,5],[6,7,8],[2,1,3]])
print(stat)
```

```python
print("Minimum:", np.min(stat))
```

```python
print("Minimum column-wise:", np.min(stat, axis=0))
```

```python
print("Minimum row-wise:", np.min(stat, axis=1))
```

```python
print("Maximum:", np.max(stat))
```

```python
print("Maximum column-wise:", np.max(stat, axis=0))
```

```python
print("Maximum row-wise:", np.max(stat, axis=1))
```

```python
print("Sum:", np.sum(stat))
```

```python
print("Sum column-wise:", np.sum(stat, axis=0))
```

```python
print("Sum row-wise:", np.sum(stat, axis=1))
```

```python
## Reorganising Arrays
```

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)
```

```python
after1 = before.reshape((8,1))
print(after1)
```

```python
after2 = before.reshape((4,2))          # printing 4 * 2 matrix
print(after2)
```

```python
after3 = before.reshape((2,2,2))        # printing 3D matrix
print(after3)             
```

```python
## Stacking 
```

```python
### Verticallly stacking 
```

```python
v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])

print(v1)
print(v2)
```

```python
v_stack = np.vstack([v1,v2])
print(v_stack)
```

```python
vvv_stack = np.vstack([v1,v2,v2,v1])
print(vvv_stack)
```

```python
### Horizontal Stack
```

```python
h1 = np.array([1,2,3,4])
h2 = np.array([5,6,7,8])

print(h1)
print(h2)
```

```python
h_stack = np.hstack([h1,h2])
print(h_stack)
```
