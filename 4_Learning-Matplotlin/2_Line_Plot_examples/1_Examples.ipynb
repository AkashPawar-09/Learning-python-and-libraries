## Working with CSV file

```python
temp = pd.read_csv("city_temperature.csv")
print(temp)
```

```python
plt.plot(temp['Month'],temp['Pune'])
plt.grid()
```

```python
plt.plot(temp['Month'],temp['Pune'] , marker = 'o', ms = 10, mec = 'r' , mfc = 'hotpink' , ls = 'dotted' , lw = '2.5')

font1 = {'family':'serif','weight': 'normal','color':'blue','size':20}
font2 = {'family':'serif','weight': 'bold','color':'darkred','size':15}

plt.title("City Temperature throughout Year", fontdict = font1)
plt.xlabel("Month" , fontdict = font2)
plt.ylabel("Temperature" , fontdict = font2)
plt.show()
# ms = Marker Size , mec = Marker Color , mfc = markerfacecolor , ls = line style , lw = line width
```

```python
plt.plot(temp['Month'],temp['Pune'] , c = 'red')
plt.plot(temp['Month'],temp['Mumbai'], c = 'green')
plt.title("City Temperature throughout Year" , loc = 'left')
plt.xlabel("Month")
plt.ylabel("Temperature")
plt.grid(axis = 'x')
```

```python
plt.plot(temp['Month'],temp['Mumbai'], c = 'green')
plt.title("City Temperature throughout Year" , loc = 'right')
plt.xlabel("Month")
plt.ylabel("Temperature")
plt.grid(color = 'green', linestyle = '--', linewidth = 0.5 , axis = 'y')
```

## Subplot

```python
# plot 1:
plt.subplot(2, 1, 1)
plt.plot(temp['Month'],temp['Pune'])
plt.title("City Temperature throughout Year" , loc = 'left')
plt.xlabel("Month")
plt.ylabel("Temperature")

# plot 2:
plt.subplot(2, 1, 2)
plt.plot(temp['Month'],temp['Mumbai'])
plt.title("City Temperature throughout Year" , loc = 'left')
plt.xlabel("Month")
plt.ylabel("Temperature")

plt.show()
```

```python
# plot 1:
plt.subplot(2, 1, 1)
plt.plot(temp['Month'],temp['Pune'])
plt.title("City Temperature throughout Year" , loc = 'left')
plt.xlabel("Month")
plt.ylabel("Temperature")

# plot 2:
plt.subplot(2, 1, 2)
plt.plot(temp['Month'],temp['Mumbai'])
plt.title("City Temperature throughout Year" , loc = 'left')
plt.xlabel("Month")
plt.ylabel("Temperature")

plt.tight_layout()
plt.show()
```

```python
# plot 1:
plt.subplot(1, 2, 1)
plt.plot(temp['Month'],temp['Pune'])
plt.title("City Temperature throughout Year")
plt.xlabel("Month")
plt.ylabel("Temperature")
plt.xticks(rotation=45)

# plot 2:
plt.subplot(1, 2, 2)
plt.plot(temp['Month'],temp['Mumbai'])
plt.title("City Temperature throughout Year")
plt.xlabel("Month")
plt.ylabel("Temperature")
plt.xticks(rotation=45)

plt.suptitle("Temperature Subplots")
plt.tight_layout()
plt.show()
```
