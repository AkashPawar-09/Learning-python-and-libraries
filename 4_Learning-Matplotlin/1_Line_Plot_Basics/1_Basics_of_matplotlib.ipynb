# Basic plots

## Importing matplotlib

```python
import matplotlib.pyplot as plt
```

## Importing numpy

```python
import numpy as np
```

```python
plt.plot([1, 2, 3, 4])    # when no axis difference ---> automatically considered as Yaxis
# If we do not specify the points on the x-axis, they will get the default values 0, 1, 2, 3 etc., depending on the length of the y-points.
```

```python
plt.plot([10, 15, 30, 55])     # Yaxis coordinates ----> 10,15,30,55
# If we do not specify the points on the x-axis, they will get the default values 0, 1, 2, 3 etc., depending on the length of the y-points.
```

```python
xpoints = np.array([0, 6])      # array having 2 points as x points
ypoints = np.array([0, 250])    # array having 2 points as y points

plt.plot(xpoints, ypoints)      # ploting x and y points
plt.show()                      # shows the plot
```

```python
plt.plot([1, 2, 3, 4], [1, 4, 9, 16])   # ploting of Xaxis,Yaxis
plt.xlabel('X axis - 1 to 4 numbers')   # label for X axis
plt.ylabel('Y axis - 1 to 4 squares')   # label for Y axis
plt.show()    # Display all open figures.

# inside plot we write Xpoints and Ypoints
```

```python
xpoints = np.array([0, 4, 6])      # array having 3 points as x points
ypoints = np.array([0, 100, 250])    # array having 3 points as y points

plt.plot(xpoints, ypoints)      # ploting x and y points
plt.show()                      # shows the plot

# Here at (4,100) there is a bend in line beacause of points we are giving to plot
```

```python
xpoints = np.array([1, 2, 6, 8])
ypoints = np.array([3, 8, 1, 10])

plt.plot(xpoints, ypoints)
plt.show()

# we want points at (1,3) , (2,8) , (6,1) , (8,10)
```
